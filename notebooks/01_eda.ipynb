# Phase 2 — EDA: Khám phá dữ liệu bán hàng (60 SKU tại CA_1)

> **Quy tắc bắt buộc**:
- Mọi phân tích mô tả và chọn mẫu **chỉ sử dụng dữ liệu giai đoạn train ($d_1 \dots d_{1577}$)**.
- Tuyệt đối không dùng thông tin giai đoạn hold-out ($d_{1578} \dots d_{1941}$).
- Biểu đồ kết xuất được lưu tại `results/eda/`.

```python
import sys
from pathlib import Path
project_root = Path("..").resolve()
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from src.data.loader import load_processed_daily
from src.data.subset import load_data_config
from src.data.eda import run_eda
```

## 1. Nạp dữ liệu và kiểm tra kích thước

```python
df = load_processed_daily("../data/processed/sales_daily.parquet")
config = load_data_config("../configs/data_config.yaml")
last_train_day = config["timeline"]["last_train_day"]

# Chỉ lấy d_1..d_1577 cho EDA mô tả
train_d = set(f"d_{i}" for i in range(1, last_train_day + 1))
train_df = df[df["d"].isin(train_d)].copy()
train_df["date"] = pd.to_datetime(train_df["date"])

print(f"Tổng số dòng train: {len(train_df):,}")
print(f"Khoảng thời gian: {train_df['date'].min().date()} đến {train_df['date'].max().date()}")
print(f"Số lượng SKU: {train_df['item_id'].nunique()}")
print(train_df['cat_id'].value_counts() // last_train_day)
```

## 2. Chạy toàn bộ EDA và xuất biểu đồ vào `results/eda/`
